The Pokémon dataset is ideal for practicing EDA because it includes a mix of numerical features (such as Attack, Defense, Speed, HP) and categorical features (such as Type 1, Type 2, and Generation). This combination allows you to apply a wide range of EDA techniques, including summary statistics, data visualizations, grouping and aggregation, correlation analysis, and comparisons across categories.

Throughout the exercise, you will explore questions such as:

## 2. Summary Statistics (4 pts)

1. Generate `df.describe()`.

```python
df.describe()
```

2. Get mean, median, and mode of Attack.

```python
mean_attack = df['Attack'].mean()
median_attack = df['Attack'].median()
mode_attack = df['Attack'].mode()[0]

print("Mean Attack:", mean_attack)
print("Median Attack:", median_attack)
print("Mode Attack:", mode_attack)
```

3. Compute 25th and 75th percentiles for HP.

```python
hp25 = df['HP'].quantile(0.25)
hp75 = df['HP'].quantile(0.75)

print("25th percentile for HP:", hp25)
print("75th percentile for HP:", hp75)
```

4. Compute standard deviation and variance of Speed.

```python
std_speed = df['Speed'].std()
var_speed = df['Speed'].var()

print("Standard Deviation of Speed:", std_speed)
print("Variance of Speed:", var_speed)
```

## 3. Filtering & Selection `(7 pts)`

Select all Pokémon with Attack > 100.

```python
high_attack = df[df['Attack'] > 100]
high_attack
```

Select all Pokémon whose primary type (Type 1) is "Fire".

```python
fire_pokemon = df[df['Type 1'] == 'Fire']
fire_pokemon
```

Select all Pokémon that are Legendary.

```python
legendary_pokemon = df[df['Legendary'] == True]
legendary_pokemon
```

Select all Pokémon that are Generation 1 AND Legendary.

```python
generation1_legendary = df[(df['Generation'] == 1) & (df['Legendary'] == True)]
generation1_legendary
```

Select all Pokémon that are Water type OR Grass type.

```python
water_pokemon = df[df['Type 1'].isin(['Water', 'Grass'])]
water_pokemon
```

Select all Pokémon that are Fire type AND Attack > 120.

```python
fire_pokemon_attack = df[(df['Type 1'] == 'Fire') & (df['Attack'] > 120)]
fire_pokemon_attack
```

Select all Pokémon whose type is in this list:
`["Dragon", "Ghost", "Dark"]`.

```python
dragon_pokemon = df[df['Type 1'].isin(["Dragon", "Ghost", "Dark"])]
dragon_pokemon
```

## 4. Categorical Exploration `(9 pts)`

Find the number of Pokémon per primary type.

```python
primary_type_counts = df['Type 1'].value_counts()
primary_type_counts
```

Find the number of Pokémon per generation.

```python
generation_pokemon_counts = df['Generation'].value_counts()
generation_pokemon_counts
```

Which type appears the most? Which appears the least?

```python
type_pokemon_most = primary_type_counts.idxmax()
type_pokemon_least = primary_type_counts.idxmin()

print("Type that appears the most:", type_pokemon_most)
print("Type that appears the least:", type_pokemon_least)
```

How many unique primary types (Type 1) exist?

```python
primary_type_counts = df['Type 1'].nunique()
primary_type_counts
```

How many unique secondary types (Type 2) exist?

```python
secondary_type_counts = df['Type 2'].nunique()
secondary_type_counts
```


Which primary types have the most dual-type combinations?

```python
dual_type_counts = df[df['Type 2'].notnull()]['Type 1'].value_counts()
dual_type_counts
```

Which type has the highest mean Attack?

```python
high_attack_type = df.groupby('Type 1')['Attack'].mean().idxmax()
high_attack_type
```

Which type has the lowest mean Defense?

```python
lowest_defense_type = df.groupby('Type 1')['Defense'].mean().idxmin()
lowest_defense_type
```


Which generation has the highest average Speed?

```python
highest_speed_generation = df.groupby('Generation')['Speed'].mean().idxmax()
highest_speed_generation
```

## 5. Groupby & Aggregation `(13 pts)`

Compute the average Attack per primary type.

```python
from numpy import average
average_attack_per_type = df.groupby('Type 1')['Attack'].mean()
average_attack_per_type
```

Compute the maximum HP per generation.


```python
maximumage_hp_per_generation = df.groupby('Generation')['HP'].max()
maximumage_hp_per_generation
```

Compute the total number of Pokémon per primary type.

```python
total_pokemon_per_type = df.groupby('Type 1').size()
total_pokemon_per_type
```

For each type, compute:

- mean Attack

- mean Defense

- mean Speed

```python
type_stats = df.groupby('Type 1')[['Attack', 'Defense', 'Speed']].mean()
type_stats
```

For each generation, compute:

- count

- mean Total

- number of Legendary Pokémon (hint: use sum on Boolean)

```python
generation_stats = df.groupby('Generation').agg({
    'Total': 'mean',
    'Legendary': 'sum'
})
generation_stats
```

Which type combination (Type 1 + Type 2) has the highest average Attack?

```python
highest_average_attack = df.groupby(['Type 1', 'Type 2'])['Attack'].mean().idxmax()
highest_average_attack
```

Which generation has the highest proportion of Legendary Pokémon?

```python
highest_proportion_legendary = df.groupby('Generation')['Legendary'].mean().idxmax()
highest_proportion_legendary
```


Which primary type has the largest variance in HP?

```python
largest_variance_hp = df.groupby('Type 1')['HP'].var().idxmax()
largest_variance_hp
```

Which primary type has the highest median Speed?

```python
highest_median_speed = df.groupby('Type 1')['Speed'].median().idxmax()
highest_median_speed
```

Group Pokémon by whether they are Legendary or not. Compare:

- mean Total

- mean Attack

- mean Defense

- mean Speed

```python
legendary_pokemon = df[df['Legendary'] == True]
non_legendary_pokemon = df[df['Legendary'] == False]
```

Show the top 5 strongest types by mean Total.

```python
strongest_types = df.groupby('Type 1')['Total'].mean().nlargest(5)
strongest_types
```

Rank generations by their average Attack.

```python
highest_average_attack = df.groupby('Generation')['Attack'].mean().sort_values(ascending=False)
highest_average_attack
```

Show the top 10 fastest Pokémon using nlargest(10, "Speed").

```python
fastest_pokemon = df.nlargest(10, 'Speed')
fastest_pokemon
```
